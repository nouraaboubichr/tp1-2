```python
import nltk, spacy, sklearn

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
import string

text = "The Cats were running QUICKLY through the beautiful gardens, and the children are playing happily! Isn't it amazing? #NLP"

text_lower = text.lower()
text_clean = text_lower.translate(str.maketrans('', '', string.punctuation))

print("Avant :", text)
print("Après :", text_clean)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

tokens = word_tokenize(text_clean)
print("Tokens :", tokens)

stop_words = set(stopwords.words('english'))
print(sorted(stop_words))

filtered = [t for t in tokens if t not in stop_words]
print("Mots restants :", filtered)

# Ajout d'un mot sans valeur sémantique
stop_words.add("isnt")
filtered = [t for t in tokens if t not in stop_words]
print("Après ajout :", filtered)

removed = (len(tokens) - len(filtered)) / len(tokens)
print(f"Proportion supprimée : {removed:.0%}")
```

```python
from nltk.stem import WordNetLemmatizer, PorterStemmer

lemmatizer = WordNetLemmatizer()
lemmas_noun = [lemmatizer.lemmatize(w) for w in filtered]
lemmas_verb = [lemmatizer.lemmatize(w, pos='v') for w in lemmas_noun]

stemmer = PorterStemmer()
for avant, lem in zip(filtered, lemmas_verb):
    print(f"{avant:12} -> lemme: {lem:10} | stem: {stemmer.stem(avant)}")
```

```python
# Version spaCy (gère automatiquement la nature grammaticale)
nlp = spacy.load("en_core_web_sm")
print([tok.lemma_ for tok in nlp(" ".join(filtered))])
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

corpus = [
    "The cats are running in the garden",
    "The children are playing in the garden",
    "Cats and children love running",
]

bow = CountVectorizer(stop_words='english')
X = bow.fit_transform(corpus)
print(pd.DataFrame(X.toarray(), columns=bow.get_feature_names_out()))

tfidf = TfidfVectorizer(stop_words='english')
T = tfidf.fit_transform(corpus)
print(pd.DataFrame(T.toarray().round(2), columns=tfidf.get_feature_names_out()))
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

phrases = [
    "The football team won the match in the stadium",
    "The movie was great and the actors played very well",
    "The team watched the movie after the match",
]

cv = CountVectorizer()
X = cv.fit_transform(phrases)

print("Mots extraits :", list(cv.get_feature_names_out()))
print("\nMatrice :\n", X.toarray())

df_bow = pd.DataFrame(X.toarray(), columns=cv.get_feature_names_out())
print("\n", df_bow)
```

```python
tv = TfidfVectorizer()
T = tv.fit_transform(phrases)

df_tfidf = pd.DataFrame(T.toarray().round(2), columns=tv.get_feature_names_out())
print(df_tfidf)

# Comparaison BoW vs TF-IDF pour la phrase 1
comparaison = pd.DataFrame({
    "BoW": df_bow.iloc[0],
    "TF-IDF": df_tfidf.iloc[0],
})
print(comparaison[comparaison["BoW"] > 0])

# Variante sans stopwords
tv2 = TfidfVectorizer(stop_words="english")
T2 = tv2.fit_transform(phrases)
print(pd.DataFrame(T2.toarray().round(2), columns=tv2.get_feature_names_out()))
```
